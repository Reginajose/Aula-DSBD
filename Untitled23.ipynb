```python
%%R
# Re-definir a função DMA para garantir que esteja disponível e robusta
dma <- function(x){
  x <- na.omit(x)
  # Retorna NA se houver 0 ou 1 ponto de dado válido para evitar erros de cálculo
  if (length(x) <= 1) return(NA)
  sum(abs(x - mean(x))) / (length(x)-1)
}

# Obter os nomes das colunas numéricas no dataset airquality
numeric_cols <- names(airquality)[sapply(airquality, is.numeric)]

# Flag para verificar se a condição é verdadeira para todas as variáveis
all_condition_true <- TRUE

# Iterar sobre cada coluna numérica
for (col_name in numeric_cols) {
  col_data <- airquality[[col_name]]

  # Calcular a variância
  var_val <- var(col_data, na.rm = TRUE)

  # Calcular o DMA
  dma_val <- dma(col_data)

  # Se algum cálculo resultar em NA (por exemplo, todos os NAs ou dados insuficientes),
  # consideramos a condição não satisfeita para essa variável.
  if (is.na(var_val) || is.na(dma_val)) {
    all_condition_true <- FALSE
    break
  }

  # Verificar a condição: variância > DMA
  if (!(var_val > dma_val)) {
    all_condition_true <- FALSE
    break # Se a condição for falsa para qualquer variável, não precisamos verificar as demais
  }
}

# Imprimir o resultado final (TRUE = 1, FALSE = 0)
cat(as.integer(all_condition_true))
```

### Implementação das funções auxiliares e da função de transformação de matriz em R

```python
%%R

# Função auxiliar para verificar se um número é primo
is_prime <- function(n) {
  if (n <= 1) return(FALSE)
  if (n == 2) return(TRUE)
  if (n %% 2 == 0) return(FALSE)
  # Check for divisibility from 3 up to sqrt(n)
  i <- 3
  while (i * i <= n) {
    if (n %% i == 0) return(FALSE)
    i <- i + 2
  }
  return(TRUE)
}

# Função auxiliar para verificar se um número é um quadrado perfeito
is_perfect_square <- function(n) {
  if (n < 0) return(FALSE)
  if (n == 0) return(TRUE)
  s <- round(sqrt(n))
  return(s * s == n)
}

# Função principal para varrer e transformar a matriz
transformar_matriz <- function(matriz) {
  matriz_original <- matriz # Guardar a matriz original para a regra do número negativo
  matriz_transformada <- matriz # Inicializar a matriz transformada

  for (i in 1:nrow(matriz)) {
    for (j in 1:ncol(matriz)) {
      valor_original <- matriz_original[i, j]
      valor_atual <- matriz_transformada[i, j]

      # Regra 1: Se o valor for um número primo, multiplique-o por 10.
      if (is_prime(valor_atual)) {
        valor_atual <- valor_atual * 10
      }

      # Regra 2: Se o valor for um quadrado perfeito, subtraia 11.
      # Caso o resultado obtido for negativo, eleve a 5ª potência.
      # Importante: A verificação de quadrado perfeito é no 'valor_atual' após a possível transformação de primo,
      # mas a instrução parece sugerir que é sobre o valor do elemento se ele for um quadrado perfeito.
      # Vou interpretar que 'Se o valor for um quadrado perfeito' se refere ao valor antes da multiplicação por 10.
      # Para evitar ambiguidade, vou aplicar as regras em ordem de prioridade. A regra de número negativo deve usar o valor original.
      # A regra do quadrado perfeito e primo não se aplicam a valores negativos.

      if (valor_original >= 0 && is_perfect_square(valor_original)) {
        temp_val <- valor_atual - 11
        if (temp_val < 0) {
          valor_atual <- temp_val^5
        } else {
          valor_atual <- temp_val
        }
      }

      # Regra 3: Se o valor for negativo na matriz original, calcule a raiz 4 de seu módulo.
      # Esta regra deve ser aplicada **apenas** aos valores que eram negativos na matriz original,
      # e não deve ser afetada pelas transformações anteriores de primo ou quadrado perfeito.
      # Se a regra 1 ou 2 já modificou o valor, ela deve ter prioridade sobre esta, a menos que especificado de outra forma.
      # A descrição 'Se o valor for negativo na matriz original' sugere que esta é uma condição `if` sobre `valor_original`.
      # Vou tratar esta regra como independente e aplicando-a se a condição for cumprida no valor original.

      if (valor_original < 0) {
        valor_atual <- abs(valor_original)^(1/4)
      }

      matriz_transformada[i, j] <- valor_atual
    }
  }
  return(matriz_transformada)
}
```

### Definição da Matriz A e cálculo da soma da diagonal principal

```python
%%R

# Matriz A
Matriz_A <- matrix(c(3, 1, -7, -8, 4, 2, -9, -4, 5, 9, 6, 10, -3, 0, -1, 7), nrow = 4, byrow = FALSE)

# Transformar Matriz A
Matriz_A_transformada <- transformar_matriz(Matriz_A)

# Obter a diagonal principal da matriz transformada
diagonal_principal <- diag(Matriz_A_transformada)

# Calcular a soma dos elementos da diagonal principal
soma_diagonal <- sum(diagonal_principal)

cat("A soma dos elementos da diagonal principal da matriz A transformada é:", soma_diagonal, "\n")
```

```python
%%R

# Matriz B
Matriz_B <- matrix(c(9, -1, -9, -3, 19, 18, 5, -6, 11, 0, -13, -16, 1, -7, 14, -5, 20, 17, -20, 10) , nrow = 4, byrow = FALSE)

# Transformar Matriz B usando a função previamente definida
Matriz_B_transformada <- transformar_matriz(Matriz_B)

# Encontrar o maior elemento em módulo na matriz transformada
maior_elemento_modulo_B <- max(abs(Matriz_B_transformada))

cat("O maior elemento, em módulo, da Matriz B transformada é:", maior_elemento_modulo_B, "\n")
```

### Quantidade de números primos nas Matrizes A, B e C juntas

```python
%%R

# Matrizes A, B e C (re-definir para garantir que estão no escopo)
Matriz_A <- matrix(c(3, 1, -7, -8, 4, 2, -9, -4, 5, 9, 6, 10, -3, 0, -1, 7), nrow = 4, byrow = FALSE)
Matriz_B <- matrix(c(9, -1, -9, -3, 19, 18, 5, -6, 11, 0, -13, -16, 1, -7, 14, -5, 20, 17, -20, 10) , nrow = 4, byrow = FALSE)
Matriz_C <- matrix(c(6, -24, -6, -27, -8, 13, -18, -3, 10, -13, 2, 0, 27, 30, 7, -11) , nrow = 4, byrow = FALSE)

# Concatenar todas as matrizes em um único vetor para varredura
todos_elementos <- c(as.vector(Matriz_A), as.vector(Matriz_B), as.vector(Matriz_C))

# Contar números primos no conjunto de todas as matrizes
primos_total <- 0
for (val in todos_elementos) {
  if (is_prime(val)) {
    primos_total <- primos_total + 1
  }
}

cat("As matrizes A, B e C juntas têm quantos números primos?:", primos_total, "\n")
```
